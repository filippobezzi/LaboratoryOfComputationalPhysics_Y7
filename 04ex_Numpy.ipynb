2\. Find the outer product of the following two vecotrs

The outer product between two vectors is defined as following:

   [[a_0*b_0  a_0*b_1 ... a_0*b_{N-1} ]

   [a_1*b_0    .

   [ ...          .
   
   [a_{M-1}*b_0            a_{M-1}*b_{N-1} ]]
   

```python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
```

Do this in the following ways:

   * Using the function outer in numpy --> "numpy.outer(a, b, out=None)" computes the outer product of two vectors "a" and ""b".
   * Using a nested for loop or list comprehension
   * Using numpy broadcasting operatoins


````python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
outer_np = np.outer(u,v)
outer_l = np.array(
    [[u[i]*v[j] for j in range(len(v))] for i in range(len(u))]
)
outer_b = u[:, np.newaxis] * v

print(outer_np == outer_l, "\n\n", outer_l == outer_b)
````

  * Implement the optimization suggested in the [sieve of Eratosthenes](https://en.wikipedia.org/wiki/Sieve_of_Eratosthenes)

````python
mask = np.ones(N + 1, dtype=bool)
print(mask, len(mask))
````

````python
import itertools

for j, i in itertools.product(range(2, int(np.sqrt(len(mask)))), range(2, N + 1)):
    if (i+1)%j == 0 and (i+1) is not j: mask[i] = False
print(mask)
````

````python
a = np.arange(1,N+2)
a[mask]
````

````python
import timeit

def eratosthenes_sieve(N):
    mask = np.ones(N + 1, dtype=bool)
    for j, i in itertools.product(range(2, int(np.sqrt(len(mask)))+1), range(2, N + 1)):
        if (i+1)%j == 0 and (i+1) is not j: mask[i] = False

    return mask

def primes(N):
    return np.arange(1,N+2)[eratosthenes_sieve(N)]

Ns = [100, 1000, 10000, 100000]
for N in Ns:
    print(f"Timing for N={N}:")
    print(primes(N))
    %timeit primes(N)
        
````
